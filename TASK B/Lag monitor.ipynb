```python
import time
from confluent_kafka import Consumer, TopicPartition

BOOTSTRAP = "localhost:9092,localhost:9094,localhost:9096"
TOPIC = "urbanpulse.traffic_signals"
GROUPS = ["traffic-signal-control-HIGH_PRIORITY", "traffic-analytics-STANDARD_PRIORITY"]

def group_lag(group_id):
    c = Consumer({"bootstrap.servers": BOOTSTRAP, "group.id": group_id, "enable.auto.commit": False})
    parts = c.list_topics(TOPIC).topics[TOPIC].partitions.keys()
    tps = [TopicPartition(TOPIC, p) for p in parts]
    committed = c.committed(tps, timeout=5)
    lag = 0
    for tp, cm in zip(tps, committed):
        low, high = c.get_watermark_offsets(tp, timeout=5)
        current = cm.offset if cm.offset and cm.offset >= 0 else low
        lag += max(high - current, 0)
    c.close()
    return lag

for _ in range(12):
    row = {g: group_lag(g) for g in GROUPS}
    print(time.strftime("%H:%M:%S"), row)
    time.sleep(5)
```
