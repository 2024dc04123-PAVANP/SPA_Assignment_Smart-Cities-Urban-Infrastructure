```python
import json, time
from confluent_kafka import Consumer

consumer_id = "2"   # "2" and "3" in the other two kernels
consumer = Consumer({
    "bootstrap.servers": "localhost:9092,localhost:9094,localhost:9096",
    "group.id": "traffic-analytics-STANDARD_PRIORITY",
    "auto.offset.reset": "latest",
})
consumer.subscribe(["urbanpulse.traffic_signals"])

while True:
    msg = consumer.poll(0.5)
    if msg is None or msg.error():
        continue
    event = json.loads(msg.value())
    time.sleep(0.5)   # simulated slow analytics pipeline -> lag builds up here
    print(f"[STANDARD-{consumer_id}] zone={event.get('zone')}")
```
